# Introducción a la Visualización de Datos

## ¿Por qué visualizar datos?

- Detectar patrones
- Comunicar hallazgos de manera efectiva
- Tomar decisiones
- Identificar errores en datos

---


# Principios de Buenas Visualizaciones

## 1. Claridad > Estética
- Evitar gráficos recargados
- Menos es más

## 2. Elegir el gráfico correcto
- Comparación de variables categoricas → Gráfico de barras / Grafico de columnas
- Distribución de una variable númerica continua → histogramas
- Relación de variables númericas → Gráfico de dispersión / scatterplot
- Composición → Gráfica de torta (con cuidado) // TreeMap
- Detectar valores atípicos de una variable númerica → Gráfico de Cajas y Bigotes

## 3. Contexto **siempre**
- Títulos claros
- Etiquetas en ejes
- Leyendas

## 4. No distorsionar datos
- No manipular escalas
- Evitar gráficos engañosos

## 5. Consistencia visual
- Colores coherentes
- Tipografías uniformes

# Hora de Practicar

```python
import pandas as pd
import matplotlib.pyplot as plt  ### Libreria Base para visualización de datos
import seaborn as sns  ### Libreria basada en Matplotlib para visualización de datos estadísticos
# Cargar dataset de ejemplo
df = sns.load_dataset("tips")

df.head()
```

```python
## Cual de los dos generos deja mas propina en promedio, hombres o mujeres?
df.groupby("sex")["tip"].mean()
```

```python
### Gráfica de propinas promedio por genero

plt.figure(figsize=(8, 4))  ## Tamaño de la gráfica
plt.title("Promedio de propinas por genero")  ## Titulo de la gráfica
plt.xlabel("Genero")  ## Etiqueta del eje X
plt.ylabel("Promedio de propina")  ## Etiqueta del eje Y
plt.grid() # Cuadricula de la gráfica
sns.barplot(data=df, x="sex", y="tip", palette="tab10")  ## Gráfica de barras con seaborn)
plt.show() # Mostrar unicamente la gráfica, sin mostrar el objeto de la gráfica
```

```python
sumatoria_propinas = df.groupby("time")["tip"].sum()
sumatoria_propinas
```

```python
df["time"].value_counts()
```

```python
# Grafico de torta
plt.pie(df["time"].value_counts(), labels=df["time"].value_counts().index, autopct='%1.1f%%')
plt.show()
```

```python
## GRafico de dispersión
plt.figure(figsize=(8, 5))
plt.title("Gráfico de dispersión de propinas vs total de la cuenta por hora del dia")
plt.xlabel("Total de la cuenta")
plt.ylabel("Propina")
sns.scatterplot(data=df, x="total_bill", y="tip", hue="time")
plt.grid()
```

```python
## Grafica de tendencia (Linea)
sns.lineplot(data=df, x="total_bill", y="tip", hue="time")
```

```python
## Gráfico de barras horizontal (de columnas)
sns.barplot(data=df,y="day",x="tip", estimator='mean', orient='y')
```

# Errores comunes

- Demasiados colores
- No etiquetar ejes
- Gráficos innecesarios
- Escalas truncadas
- Uso incorrecto de pie charts


# Uso del Color

## Buenas prácticas:

- Usar colores con propósito
- No usar más de 3-5 colores
- Usar contraste para resaltar
- Evitar colores innecesarios

```python
# Ejemplo de uso de color intencional

plt.figure(figsize=(8,5))
sns.barplot(x='day', y='total_bill', data=df, palette='Set2')

plt.title("Uso de paleta monocromática")
plt.show()
```
