## 1. Conectarse a la base de datos de interes (USE)

```sql
USE sakila;
```

## 2. Consulta básica sobre una tabla (SELECT)

Recuerde que el * significa que esoy consultando todas las columnas de la tabla 

```sql
SELECT * FROM actor; 
```

Si quiero consultar columnas especificas, puedo listarlas entre el comando SELECT y FROM

```sql
SELECT first_name, last_name FROM actor; 
```

Es posible utilizar el comando **LIMIT** para consultar solo una cantidad especifica de filas

```sql
SELECT * FROM actor LIMIT 10; 
```

## 3. Hacer filtros en las consultas (WHERE) 

El comando WHERE me sirve para filtrar las filas basandome en una condicion de verdad

```sql
SELECT * FROM actor 
WHERE first_name = 'JOHNNY';  
```

Dependiendo del tipo de dato de la columna, podemos hacer consultas a nivel numerico con los simbolos >, < , >= , <=

```sql
SELECT * FROM film
WHERE length >= 120;
```

Para las columnas de VARCHAR (Texto), puedo hacer busquedas por similitud de caracteres, usando el comando LIKE ylos caracteres de escape como el %

```sql
# Consultar todas las peliculas que empiezan por la letra A
SELECT * FROM film
WHERE title LIKE 'A%';

```

Puedo hacer filtros sobre columnas calculadas como por ejemplo la funcion LENGTH

```sql
# Consultar todas las peliculas que tienen menos de 10 caracteres en el titulo
SELECT title, LENGTH(title) FROM film
WHERE LENGTH(title) < 10;
```

## 4. Organizar los datos resultantes de la consulta (ORDER BY)

Puedo ordenar los datos de la consulta de manera ascendente

```sql
SELECT first_name, last_name
FROM actor
ORDER BY first_name ASC;
```

Puedo ordenar los datos de la consulta de manera Descendente

```sql
SELECT first_name, last_name
FROM actor
ORDER BY first_name DESC;
```

Puedo ordenar los datos de la consulta basandome en multiples parámetros al tiempo

```sql
## Ordenar los datos por nombre ascendente y apellido descendente
SELECT first_name, last_name
FROM actor
ORDER BY first_name ASC, last_name DESC ;
```

## 6. Actualizar datos existentes (UPDATE)

 Actualizar del actor recien ingresado. Puedo actualizar un campo o varios campos al tiempo

```sql
UPDATE actor
SET 
first_name = 'SYLVESTER',
last_name = 'STALLONE'
WHERE actor_id = '201';
```

## 7. Borrar datos existentes (DELETE)

Borrar el actor recien ingresado. IMPORTANTE **SIEMPRE USAR EL WHERE**  

``` sql

WHERE actor_id = 201;
```

Puede usar Transaction Control Language (TCL) dentro de MYSQL para manejar el guardado de las transacciones que se realizan en la BD

```sql
START TRANSACTION; 
DELETE FROM actor
WHERE actor_id = 201;

SELECT * FROM Actor:

## SI no estoy conforme con ese cambio en la BD
ROLLBACK;

## SI estoy conforme con ese cambio en la BD
COMMIT;

```

# Agregaciones (GROUP BY)

Las funciones de agregación me sirven para resumir datos:

- COUNT() → contar
- SUM() → sumar
- AVG() → promedio
- MAX() → máximo
- MIN() → mínimo

Puedo aplicar funciones de agregación a los campos que tengo en la tabla de la siguiente manera:

```sql
SELECT COUNT(first_name) FROM actor;
```

Tambien puedo segmentar las agregacion de los datos por una variable categorica, usando la palabra clave GROUP BY

```sql
SELECT rating, SUM(rental_duration) FROM film
GROUP BY rating;
```

## Uso de HAVING 
Funciona como filtro despues de aplicar una funcion de agregación

```sql
## Muestreme los ratings de peliculas con un promedio de duración mayor a 115
SELECT rating, avg(length) FROM film
# WHERE length > 115 NO VA EL WHERE
GROUP BY rating
HAVING avg(length) > 115;
```

# Relacionar Tablas (JOIN)


Puedo relacionar diferentes tablas por medio de sus llaves primarias y foraneas, usando la palabra JOIN

```sql
SELECT actor.first_name, actor.last_name, film.title FROM film_actor
JOIN film ON film_actor.film_id = film.film_id
JOIN actor ON film_actor.actor_id = actor.actor_id
```

Puedo combinar estos JOIN de relacion de tablas con las funciones de agregacion, si es necesario
```sql
SELECT film.title , COUNT(actor.first_name)FROM film_actor
JOIN film ON film_actor.film_id = film.film_id
JOIN actor ON film_actor.actor_id = actor.actor_id
GROUP BY film.title 
```
