# ¿Qué es y para qué sirve el EDA?

El Análisis Exploratorio de Datos (EDA) es el proceso de **examinar, resumir y visualizar** un dataset para entender su estructura, calidad y patrones antes de hacer modelos o conclusiones.

## ¿Para qué sirve?

- Detectar errores y problemas en los datos
- Entender la distribución de las variables
- Identificar relaciones entre variables
- Generar hipótesis
- Guiar decisiones de:
  - limpieza
  - transformación
  - modelado

## Comunmente hacemos esto en el EDA

Inspección inicial
- Tamaño del dataset (filas, columnas) ---> df.shape
- Tipos de datos (tipo de dato hay en cada columna) --> df.info / df.dtypes / df.columns
- Primeras filas (conocer como se ven los datos) --> df.head()

Resumen Estadístico  --> df.describe()
- Promedio, mediana, desviación
- Mínimos y máximos

Valores faltantes  --> df.isnull() / df.info()
- Cuántos hay?
- Dónde están?

Duplicados ---> df.duplicated()
- Filas repetidas
- Llave primaria repetida

Categorias únicas ---> df['columna_categorica'].value_counts()
- Revision de valores distintos en las variables categoricas
- A partir de esto puedo hacer normalización de los datos

Filtrado y agrupación  ---->df.groupby()
- Segmentar datos
- Comparar grupos

## Análisis Univariado (una variable)

- Variables numéricas
  - Distribución (Histograma)
  - Outliers (Cajas y bigotes)
  - Rango de valores (Cajas y bigotes)
- Variables categóricas
  - Frecuencia de categorías (Gráfico de barras)
  - Categorías dominantes (Gráfico de barras)

## Análisis Multivariado (relaciones)

- Numérica vs Numérica
  - Correlación (Dispersión)
  - Tendencias (Gráfico de linea)

- Categórica vs Numérica
  - Comparación entre grupos (Gráfico de barras)
