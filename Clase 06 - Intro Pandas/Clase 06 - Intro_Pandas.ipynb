## 3. Filtros y Selección (Preguntarle a los Datos)
### A. Seleccionar Columnas

```python
# Seleccionar una sola columna (obtenemos una Serie)
df_ventas['Videojuego'].head()
```

```python
# Seleccionar múltiples columnas (pasamos una lista dentro de los corchetes)
columnas_interes = ['Videojuego', 'Plataforma', 'Unidades_Vendidas']
df_ventas[columnas_interes].head()
```

### B. Filtrar Filas con Máscaras Booleanas (Paso a Paso)
Queremos ver solo las ventas donde se vendieron más de 20 unidades.

```python
df_ventas.head()
```

```python
# Paso 1: Crear la condición (devuelve True o False por cada fila)
condicion = df_ventas['Unidades_Vendidas'] <= 20
condicion.head()
```

```python
# Paso 2: Pasar esa condición al DataFrame original
df_grandes_ventas = df_ventas[condicion]
df_grandes_ventas
```

Ventas en latinoamerica

```python
mascara = df_ventas['Region'] == "Latinoamerica"
mascara.head()
```

Aplico la mascara y funciones de ordenamiento

```python
df_ventas[mascara].sort_values(by=["Plataforma", "Unidades_Vendidas"], ascending=False)
```

Filtro con multilples condiciones

```python
condicion1 = df_ventas['Unidades_Vendidas'] > 20
condicion2 = df_ventas['Region'] == "Latinoamerica"

df_ventas[condicion1 & condicion2] 
```

### C. Filtros Múltiples
Usamos `&` para "Y" (ambas deben cumplirse) y `|` para "O" (al menos una debe cumplirse).
**Importante:** Cada condición debe ir entre paréntesis `()`.

```python
# Filtrar ventas de PS5 Y que sean de la región 'Latinoamerica'
filtro_combinado = (df_ventas['Plataforma'] == 'PS5') & (df_ventas['Region'] == 'Latinoamerica')
df_ventas[filtro_combinado]
```

--- 
## 4. Transformaciones Vectorizadas 
En Pandas, no usamos ciclos `for` o `while` para operar celdas. Hacemos operaciones directas entre columnas.

```python
# Crear una columna calculada directamente
df_ventas['Ingreso_Estimado'] = df_ventas['Precio_Unitario'] * df_ventas['Unidades_Vendidas']
df_ventas.head()
```

### Manipulación de Texto Vectorizado con `.str`
Si queremos limpiar o extraer partes de un texto, usamos los métodos de string nativos de Pandas.

```python
# Ejemplo: Convertir la región a mayúsculas sostenidas de una sola vez
df_ventas['Region_Mayus'] = df_ventas['Region'].str.upper()
df_ventas[['Region', 'Region_Mayus']].head()
```

```python
df_ventas
```

--- 
## 5. Agrupaciones (`groupby`)
El método `groupby` nos permite resumir información por categorías siguiendo la lógica: Dividir -> Aplicar -> Combinar.

```python
# Pregunta: ¿Cuántas unidades totales se vendieron por cada Plataforma?
# Agrupamos por 'Plataforma', seleccionamos 'Unidades_Vendidas' y aplicamos la suma (.sum())
ventas_por_consola = df_ventas.groupby('Plataforma')['Unidades_Vendidas'].sum()
ventas_por_consola
```

--- 
## 6. Exportar Resultados
Una vez terminado nuestro análisis, podemos guardarlo directamente en un nuevo archivo Excel o CSV.

```python
# Guardamos el resumen anterior en un CSV
ventas_por_consola.to_csv('resumen_ventas_plataformas.csv', header=True)
print("¡Archivo exportado con éxito!")
```
